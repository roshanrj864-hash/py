```python
ACTIVITY
1. Identify the Components
1.1. What are the inputs?

Three numerical quiz scores entered by the user (quiz_1, quiz_2, quiz_3).

A string input from the user to decide whether to continue the loop (choice).

1.2. What is the process?

Initialize a variable to control the loop.

Start a loop that runs as long as the user wants to continue.

Prompt the user to input the scores for three quizzes.

Calculate the sum of the three quiz scores and divide by 3 to find the average.

Check if the calculated average is greater than or equal to 50.

If the average is 50 or higher, determine the status as "PASS"; otherwise, determine the status as "FAIL".

Ask the user if they want to calculate another student's average to update the loop control variable.

End the program when the user chooses to stop.

1.3. What is the output?

A message displaying the student's status: "PASS" or "FAIL".

A final termination message: "Program Ended".

2. Design the Algorithm
2.1. Pseudocode

Plaintext
SET choice = 'y'

WHILE choice == 'y'
    INPUT quiz_1
    INPUT quiz_2
    INPUT quiz_3

    COMPUTE average = (quiz_1 + quiz_2 + quiz_3) / 3

    IF average >= 50 THEN
        PRINT "PASS"
    ELSE
        PRINT "FAIL"
    END IF

    INPUT choice ("Continue? (y/n): ")
END WHILE

PRINT "Program Ended"
2.2. Identify the algorithm structure:

2.2.1. Sequence: This is the top-to-bottom, step-by-step execution of the program. For example, within your loop, the program logically follows a sequence: first it asks for quiz_1, then quiz_2, then quiz_3, then it calculates the average. It must happen in this exact order to work.

2.2.2. Selection: This is the decision-making part of the code. In your program, the if/else statement (if average >= 50:) represents the selection structure because the program chooses to execute one block of code (print("PASS")) or another (print("FAIL")) based on a specific condition.

2.2.3. Iteration: This is the repeating or looping part of the code. In your program, the while choice == 'y': statement is the iteration structure. It allows the program to repeatedly run the sequence of asking for grades and determining pass/fail until the user changes the choice variable.
```
